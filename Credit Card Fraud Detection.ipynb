CONVERTING DATA INTO STRING, INTEGER, FLOAT

```python
df["transaction_id"] = df["transaction_id"].astype("int64")
df["customer_id"] = df["customer_id"].astype("int64")
df["customer_name"] = df["customer_name"].astype(str)
df["country"] = df["country"].astype(str)

df["V1"] = df["V1"].astype("float")
df["V2"] = df["V2"].astype("float")
df["V3"] = df["V3"].astype("float")
df["V4"] = df["V4"].astype("float")
df["V5"] = df["V5"].astype("float")
df["V6"] = df["V6"].astype("float")
df["V7"] = df["V7"].astype("float")
df["V8"] = df["V8"].astype("float")
df["V9"] = df["V9"].astype("float")
df["V10"] = df["V10"].astype("float")
df["V11"] = df["V11"].astype("float")
df["V12"] = df["V12"].astype("float")
df["V13"] = df["V13"].astype("float")
df["V14"] = df["V14"].astype("float")
df["V15"] = df["V15"].astype("float")
df["V16"] = df["V16"].astype("float")
df["V17"] = df["V17"].astype("float")
df["V18"] = df["V18"].astype("float")
df["V19"] = df["V19"].astype("float")
df["V20"] = df["V20"].astype("float")
df["V21"] = df["V21"].astype("float")
df["V22"] = df["V22"].astype("float")
df["V23"] = df["V23"].astype("float")
df["V24"] = df["V24"].astype("float")
df["V25"] = df["V25"].astype("float")
df["V26"] = df["V26"].astype("float")
df["V27"] = df["V27"].astype("float")
df["V28"] = df["V28"].astype("float")

df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
df["Amount"] = df["Amount"].fillna(0).astype("int64")
df["Class"] = pd.to_numeric(df["Class"], errors="coerce")
df["Class"] = df["Class"].fillna(0).astype("int64")
```

```python
df["Time"] = pd.to_numeric(df["Time"], errors="coerce")

df["Time"] = df["Time"].fillna(0)

df["Time_Formatted"] = pd.to_timedelta(
    df["Time"] % 86400,
    unit="s"
).astype(str)

df["Hour"] = ((df["Time"] // 3600) % 24).astype("int64")
```

```python
df["customer_name"] = df["customer_name"].str.strip()
df["country"] = df["country"].str.strip()

df["customer_name"] = df["customer_name"].str.upper()
df["country"] = df["country"].str.upper()
```

REMOVING THE TIME FROM THE DATA AND RENAMING THE DATA 

```python
df = df.drop(columns=["Time"])
```

```python
df = df.rename(columns={"Time_Formatted": "Time"})
```

FILL THE MISSING VALUES

```python
df["customer_name"] = df["customer_name"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
```

```python
df["Amount"] = df["Amount"].fillna(df["Amount"].median())
df["Amount"] = df["Amount"].fillna(df["Amount"].median())
```

EXPLORATORY DATA ANALYSIS

How many transactions are in the dataset ?

```python
total_transactions = df.shape[0]
print("Total Transactions:",total_transactions)
```

How many transactions are normal and fraud ?

```python
normal_transactions = (df["Class"] == 0).sum()
fraud_transactions = (df["Class"] == 1).sum()
print("Normal Transactions:", normal_transactions)
print("Fraud Transactions:", fraud_transactions)
```

What are the total percentage of fraud transactions ?

```python
fraud_percentage = (fraud_transactions / total_transactions) * 100
print("Fraud Transactions:", fraud_transactions)
print("Fraud Percentage:", round(fraud_percentage, 2), "%")
```

What is the total transactions value ?

```python
total_amount = df["Amount"].sum()
print("Total Transactions Amount:", total_amount)
```

How many transactions is associated with fraud transactions ?

```python
total_fraud_amount = df.loc[
       df["Class"] == 1,
       "Amount"
].sum()
print("Total Fraud Amount:", total_fraud_amount)
```

What is the total average transactions value ?

```python
average_amount = df["Amount"].mean()
print("Average Transactions Amount:", average_amount)
```

How many average transactions is associated with fraud transactions ?

```python
average_fraud_amount = df.loc[
       df["Class"] == 1,
       "Amount"
].mean()
print("Average Fraud Amount:", average_fraud_amount)
```

Which hour has the highest number of fraud transactions ?

```python
fraud_by_hour = (
    df[df["Class"] == 1]
    .groupby("Hour")
    .size()
    .reset_index(name="Fraud_Count")
    .sort_values("Fraud_Count", ascending=False)
)
print(fraud_by_hour)
```

Which hour has the highest rate of fraud transactions ?

```python
hourly_fraud_rate = (
    df.groupby("Hour")["Class"]
    .agg(["count", "sum"])
    .reset_index()
)

hourly_fraud_rate["Fraud_Rate"] = (
    hourly_fraud_rate["sum"] /
    hourly_fraud_rate["count"] * 100
)

hourly_fraud_rate = hourly_fraud_rate.sort_values(
    "Fraud_Rate",
    ascending=False
)

print(hourly_fraud_rate)
```

What are the average V1-V28 values for normal and fraud transactions?

```python
v_columns = [col for col in df.columns if col.startswith("V")]

df.groupby("Class")[v_columns].mean()
```

Which V feature has the largest difference between normal and fraud?

```python
feature_mean = df.groupby("Class")[v_columns].mean()

feature_difference = (
    feature_mean.loc[1] - feature_mean.loc[0]
).abs().sort_values(ascending=False)

feature_difference
```

What are the standard deviations of V1-V28 for fraud transactions?

```python
df[df["Class"] == 1][v_columns].std().sort_values(ascending=False)
```

DATA VISUALIZATION USING MATPLOTLIB

How many transactions are normal and fraud ?

```python
df["Class"].value_counts().plot(kind="bar")

plt.title("Normal vs Fraud Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.xticks([0, 1], ["Normal", "Fraud"], rotation=0)
plt.show()
```

What is the percentage of transactions are fraud?

```python
class_count = df["Class"].value_counts()
plt.pie(
    class_count,
    labels=["Normal", "Fraud"],
    autopct="%1.1f%%"
)
plt.title("Normal vs Fraud Transaction Percentage")
plt.show()
```

Which hours have the highest number of fraud transactions?

```python
fraud_by_hour = (
    df[df["Class"] == 1]
    .groupby("Hour")
    .size()
)
fraud_by_hour.sort_index().plot(kind="line", marker="o")
plt.title("Fraud Transactions by Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Fraud Transactions")
plt.show()
```

What does the fraud transactions amount distribution look like ?

```python
fraud_amount = df.loc[
    df["Class"] == 1,
    "Amount"
]
plt.hist(fraud_amount, bins=30)
plt.title("Fraud Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.show()
```

Is the average transaction amount different between normal and fraud ?

```python
average_amount = df.groupby("Class")["Amount"].mean()

average_amount.plot(kind="bar")

plt.title("Average Transaction Amount")
plt.xlabel("Transaction Type")
plt.ylabel("Average Amount")
plt.xticks([0, 1], ["Normal", "Fraud"], rotation=0)
plt.show()
```

Which V features differ most between normal and fraud?

```python
v_columns = [col for col in df.columns if col.startswith("V")]

feature_mean = df.groupby("Class")[v_columns].mean()

feature_difference = (
    feature_mean.loc[1] - feature_mean.loc[0]
).abs().sort_values(ascending=False)

feature_difference.plot(kind="bar")

plt.title("Difference in V1-V28 Between Normal and Fraud")
plt.xlabel("Feature")
plt.ylabel("Absolute Mean Difference")
plt.show()
```
